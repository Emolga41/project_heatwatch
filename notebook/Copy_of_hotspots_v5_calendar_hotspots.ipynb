```python
def conf_class(c):
    if isinstance(c, str):
        return {"l": "low", "n": "nominal", "h": "high"}.get(c.strip().lower())
    if c < 30:  return "low"
    if c <= 80: return "nominal"
    return "high"

def clean(d, sensor):
    d = d.copy()
    if "type" in d.columns:
        d = d[d["type"] == 0].drop(columns="type")
    d["conf_class"] = d["confidence"].map(conf_class)
    d = d[d["conf_class"].isin(["nominal", "high"])]
    d["sensor"] = sensor
    return d.drop(columns="confidence")

modis_archive_c = clean(modis_archive, "MODIS")
modis_new_c     = clean(modis_new, "MODIS")
modis_nrt_c     = clean(modis_nrt, "MODIS")
viirs_archive_c = clean(viirs_archive, "VIIRS")
viirs_nrt_c     = clean(viirs_nrt, "VIIRS")
```

```python
def fit_ratios(df, mcol, by):
    overall = df["viirs"].sum() / df[mcol].sum()
    if not by:
        return overall
    g = df.groupby(by)[["viirs", mcol]].sum()
    return (g["viirs"] / g[mcol]).replace([np.inf, -np.inf], np.nan).fillna(overall)

def predict(df, mcol, k, by):
    if not by:
        return df[mcol] * k
    out = df.merge(k.rename("k").reset_index(), on=by, how="left")
    return (out[mcol] * out["k"]).values

def score(by, mcol):
    k = fit_ratios(learn, mcol, by)
    t = test.assign(pred=predict(test, mcol, k, by)).groupby("month")[["pred", "viirs"]].sum()
    err  = (t.pred - t.viirs).abs().sum() / t.viirs.sum()
    bias = (t.pred.sum() - t.viirs.sum()) / t.viirs.sum()
    return round(err, 3), round(bias, 3)
```

```python
def attach(d, k, by):
    d = d.copy()
    d["moy"] = d.acq_date.dt.month
    if not by:
        d["w"] = float(k)
    else:
        d = d.merge(k.rename("w").reset_index(), on=by, how="left")
    return d

cut_terra = pd.Timestamp("2002-07-04")
cut_viirs = pd.Timestamp("2012-02-01")

old = modis_all[modis_all.acq_date < cut_viirs]
old_t = attach(old[old.acq_date <  cut_terra], k_terra, BY_TERRA); old_t["source"] = "MODIS-adjusted (Terra only)"
old_b = attach(old[old.acq_date >= cut_terra], k_all,   BY_ALL);   old_b["source"] = "MODIS-adjusted"

new = viirs_all[viirs_all.acq_date >= cut_viirs].copy();  new["w"] = 1.0;  new["source"] = "VIIRS"
```

```python
def cell_key(d, res=0.01): #to buang the nrt extras
    return (np.floor(d.latitude / res).astype(int).astype(str) + "_" +
            np.floor(d.longitude / res).astype(int).astype(str))

va = viirs_archive[viirs_archive.confidence.map(conf_class).isin(["nominal", "high"])].copy()
va["cell"] = cell_key(va)
hist = va[va.acq_date <  "2026-01-01"]
chk  = va[va.acq_date >= "2026-01-01"]

for min_days in [1, 3, 5, 10]:
    nv = hist[hist["type"] != 0].groupby("cell").acq_date.nunique()
    st = set(nv[nv >= min_days].index)
    removed = chk.cell.isin(st)
    print(min_days, "| non-veg caught:", round(removed[chk["type"] != 0].mean(), 2),
          "| real fires wrongly removed:", round(removed[chk["type"] == 0].mean(), 2))
```

```python
def backtest(mcol, k, by):
    ev = m[(m.month >= "2012-02-01") & (m.month <= "2026-06-01")].copy()
    ev["pred"] = predict(ev, mcol, k, by)
    mon = ev.groupby("month")[["pred", "viirs"]].sum()
    mon["period"] = np.where(mon.index <= "2018-12-01", "learn",
                     np.where(mon.index <= "2020-12-01", "test", "drift"))
    return mon

def metrics(d):
    y, p = d["viirs"], d["pred"]
    e = p - y
    mse = (e ** 2).mean()
    busy = y >= 100
    return {
        "months": len(d),
        "accuracy_%": 100 * (1 - e.abs().sum() / y.sum()),
        "MSE": mse,
        "RMSE": np.sqrt(mse),
        "MAE": e.abs().mean(),
        "R2": 1 - (e ** 2).sum() / ((y - y.mean()) ** 2).sum(),
        "correlation": np.corrcoef(y, p)[0, 1],
        "bias_%": 100 * e.sum() / y.sum(),
        "within_25%_busy": 100 * ((e.abs() / y)[busy] <= 0.25).mean(),
    }


k_all_val   = fit_ratios(learn, "modis_all",   BY_ALL)
k_terra_val = fit_ratios(learn, "modis_terra", BY_TERRA)
mon_all   = backtest("modis_all",   k_all_val,   BY_ALL)
mon_terra = backtest("modis_terra", k_terra_val, BY_TERRA)

rows_m = []
for label, mon in [("Terra + Aqua", mon_all), ("Terra only", mon_terra)]:
    big = mon.viirs.idxmax()
    subsets = {
        "learn 2012-2018 (ratio built here)": mon[mon.period == "learn"],
        "learn, without biggest month":       mon[mon.period == "learn"].drop(big, errors="ignore"),
        "TEST 2019-2020 (unseen)":            mon[mon.period == "test"],
        "2021-2026 MODIS drift (not used)":   mon[mon.period == "drift"],
        "all 2012-2020":                      mon[mon.period != "drift"],
    }
    for name, d in subsets.items():
        rows_m.append({"modis_set": label, "period": name, **metrics(d)})

pd.set_option("display.width", 250); pd.set_option("display.max_columns", 30)
metrics_df = pd.DataFrame(rows_m)
print(metrics_df.round(2).to_string(index=False))
```

```python
from sklearn.linear_model import LinearRegression, PoissonRegressor
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error
from sklearn.inspection import permutation_importance

LEARN = ("2012-02-01", "2018-12-01")
TEST  = ("2019-01-01", "2020-12-01")
GRID  = 0.25
SMOOTH = 20

KEYS = ["ix", "iy", "month"]

def to_grid(d, res, start, end):
    d = d[(d.month >= start) & (d.month <= end)].copy()
    d["ix"] = np.floor(d.longitude / res).astype(int)
    d["iy"] = np.floor(d.latitude  / res).astype(int)
    return d

def modis_features(modis, res, start, end, extra_cells=None):

    mo = to_grid(modis, res, start, end)
    mo["is_terra"] = (mo.satellite == "Terra").astype(int)
    mo["is_night"] = (mo.daynight == "N").astype(int)
    mg = mo.groupby(KEYS).agg(modis_n=("frp", "size"), modis_terra_n=("is_terra", "sum"),
                              modis_night_n=("is_night", "sum"), frp_sum=("frp", "sum"),
                              frp_mean=("frp", "mean"), frp_max=("frp", "max")).reset_index()

    cells = pd.concat([mg[["ix", "iy"]]] + ([extra_cells[["ix", "iy"]]] if extra_cells is not None else [])).drop_duplicates()
    months = pd.DataFrame({"month": pd.date_range(start, end, freq="MS")})
    p = cells.merge(months, how="cross").merge(mg, on=KEYS, how="left").fillna(0)

    src = p[KEYS + ["modis_n", "frp_sum"]]
    shifted = [src.assign(ix=src.ix + dx, iy=src.iy + dy)
               for dx in (-1, 0, 1) for dy in (-1, 0, 1) if (dx, dy) != (0, 0)]
    nb = pd.concat(shifted).groupby(KEYS)[["modis_n", "frp_sum"]].sum().add_prefix("nb_").reset_index()
    p = p.merge(nb, on=KEYS, how="left").fillna(0)

    p = p.sort_values(KEYS).reset_index(drop=True)
    g = p.groupby(["ix", "iy"])
    for c in ["modis_n", "frp_sum"]:
        p[c + "_prev"] = g[c].shift(1).fillna(0)

    p["moy"]    = p.month.dt.month
    p["year"]   = p.month.dt.year
    p["lat"]    = (p.iy + 0.5) * res
    p["lon"]    = (p.ix + 0.5) * res
    p["borneo"] = (p.lon >= 105).astype(int)
    p["frp_per_hotspot_nb"] = (p.nb_frp_sum / p.nb_modis_n.replace(0, np.nan)).fillna(0)
    return p

def build_panel(res=GRID, modis=modis_all, viirs=viirs_all, learn_end=LEARN[1]):
    vi = to_grid(viirs, res, LEARN[0], TEST[1])
    vg = vi.groupby(KEYS).agg(viirs=("frp", "size"), viirs_frp_sum=("frp", "sum")).reset_index()
    p = modis_features(modis, res, LEARN[0], TEST[1], extra_cells=vg)
    p = p.merge(vg, on=KEYS, how="left")
    p[["viirs", "viirs_frp_sum"]] = p[["viirs", "viirs_frp_sum"]].fillna(0)
    p["period"] = np.where(p.month <= learn_end, "learn", "test")

    L  = p[p.period == "learn"]
    k0 = L.viirs.sum() / L.modis_n.sum()
    cy = L.groupby(["ix", "iy", "year"]).agg(v=("viirs", "sum"), m=("modis_n", "sum"), n=("viirs", "size"))
    ct = cy.groupby(level=[0, 1]).sum()
    full = pd.DataFrame({"cell_ratio": (ct.v + SMOOTH * k0) / (ct.m + SMOOTH),
                         "cell_viirs_rate": ct.v / ct.n}).reset_index()
    loo = cy.join(ct, on=["ix", "iy"], rsuffix="_tot")
    loo = pd.DataFrame({"r_loo": (loo.v_tot - loo.v + SMOOTH * k0) / (loo.m_tot - loo.m + SMOOTH),
                        "rate_loo": (loo.v_tot - loo.v) / (loo.n_tot - loo.n)}).reset_index()
    p = p.merge(full, on=["ix", "iy"], how="left").merge(loo, on=["ix", "iy", "year"], how="left")
    is_learn = p.period == "learn"
    p["cell_ratio"]      = np.where(is_learn, p.r_loo, p.cell_ratio)
    p["cell_viirs_rate"] = np.where(is_learn, p.rate_loo, p.cell_viirs_rate)
    p = p.drop(columns=["r_loo", "rate_loo"])
    p["cell_ratio"]      = p.cell_ratio.fillna(k0)
    p["cell_viirs_rate"] = p.cell_viirs_rate.fillna(0)
    p["modis_x_ratio"]   = p.modis_n * p.cell_ratio
    return p, k0

panel, k0 = build_panel()
print(f"grid {GRID}°: {panel[['ix','iy']].drop_duplicates().shape[0]} cells x "
      f"{panel.month.nunique()} months = {len(panel):,} rows | national ratio k0 = {k0:.3f}")
panel.head()
```

```python
from sklearn.metrics import mean_absolute_error

tL = panel[panel.period == "learn"].assign(moy=lambda d: d.month.dt.month, reg=lambda d: np.where(d.lon < 105, "Peninsular", "Borneo"))
tT = panel[panel.period == "test"].assign(moy=lambda d: d.month.dt.month, reg=lambda d: np.where(d.lon < 105, "Peninsular", "Borneo"))
def t_ratio(d, by):
    g = d.groupby(by)[["viirs", "modis_terra_n"]].sum()
    return (g.viirs / g.modis_terra_n).replace([np.inf, -np.inf], np.nan)
kt0, kt_m, kt_rm = tL.viirs.sum() / tL.modis_terra_n.sum(), t_ratio(tL, "moy"), t_ratio(tL, ["reg", "moy"])
c1 = tL.groupby(["ix", "iy"])[["viirs", "modis_terra_n"]].sum().reset_index()
c1["r_c"] = (c1.viirs + SMOOTH * kt0) / (c1.modis_terra_n + SMOOTH)
c2 = tL.groupby(["ix", "iy", "moy"])[["viirs", "modis_terra_n"]].sum().reset_index()
c2["r_cm"] = (c2.viirs + SMOOTH * c2.moy.map(kt_m)) / (c2.modis_terra_n + SMOOTH)
tT = tT.merge(c1[["ix", "iy", "r_c"]], on=["ix", "iy"], how="left").merge(c2[["ix", "iy", "moy", "r_cm"]], on=["ix", "iy", "moy"], how="left")
tT["r_c"], tT["r_cm"] = tT.r_c.fillna(kt0), tT.r_cm.fillna(tT.moy.map(kt_m))

METHODS = {
    "single Terra k (1 number)":  (kt0 * tT.modis_terra_n,                                     "#9e9e9e", ":"),
    "per month (12) - used now":  (tT.modis_terra_n * tT.moy.map(kt_m).fillna(kt0),            "#d62728", "-"),
    "per region x month (24)":    (tT.modis_terra_n * kt_rm.reindex(pd.MultiIndex.from_arrays([tT.reg, tT.moy])).fillna(kt0).to_numpy(), "#ff7f0e", "--"),
    "per cell (1 per square)":    (tT.modis_terra_n * tT.r_c,                                  "#1f77b4", "-."),
    "per cell x month":           (tT.modis_terra_n * tT.r_cm,                                 "#2ca02c", "--"),
}
actual = tT.groupby("month").viirs.sum()
monthly = pd.DataFrame({n: tT.assign(p=np.asarray(p, float)).groupby("month").p.sum() for n, (p, _, _) in METHODS.items()})
ape = 100 * monthly.sub(actual, axis=0).abs().div(actual, axis=0)
score = pd.DataFrame({"MAE": [mean_absolute_error(actual, monthly[n]) for n in METHODS],
                      "median APE %": ape.median().to_numpy(),
                      "bias %": (100 * (monthly.sum() / actual.sum() - 1)).to_numpy()}, index=list(METHODS)).sort_values("MAE")
best = score.index[0]

fig, ax = plt.subplots(3, 1, figsize=(13, 11), gridspec_kw={"height_ratios": [2.2, 1.3, 1]})
ax[0].plot(actual.index, actual, color="black", lw=2.6, marker="o", ms=4, label="ACTUAL VIIRS", zorder=5)
for n, (_, col, ls) in METHODS.items():
    ax[0].plot(monthly.index, monthly[n], color=col, ls=ls, lw=2.4 if n == best else 1.4, alpha=1 if n == best else 0.85,
               label=f"{n}  (MAE {score.loc[n, 'MAE']:.0f}, median APE {score.loc[n, 'median APE %']:.0f}%)" + ("  <- closest" if n == best else ""))
ax[0].set_ylabel("hotspots per month (Malaysia)")
ax[0].set_title("Terra-only MODIS converted to VIIRS: each method vs actual VIIRS (unseen test months 2019-2020)")
ax[0].legend(fontsize=8.5, loc="upper left")
for n, (_, col, ls) in METHODS.items():
    ax[1].plot(ape.index, 100 * (monthly[n] - actual) / actual, color=col, ls=ls, lw=2.2 if n == best else 1.2, marker=".")
ax[1].axhline(0, color="black", lw=0.8); ax[1].axhspan(-20, 20, color="grey", alpha=0.1)
ax[1].set_ylabel("error % (+ too high, - too low)"); ax[1].set_title("Monthly error of each method (grey band = within +/-20%)")
s = score.iloc[::-1]
ax[2].barh(s.index, s.MAE, color=[METHODS[n][1] for n in s.index])
for y, (m, a, b) in enumerate(zip(s.MAE, s["median APE %"], s["bias %"])):
    ax[2].text(m, y, f"  MAE {m:.0f} | median APE {a:.0f}% | bias {b:+.1f}%", va="center", fontsize=8.5)
ax[2].set_xlabel("MAE of monthly Malaysian total (lower = better)"); ax[2].set_xlim(0, s.MAE.max() * 1.6)
for a_ in ax: a_.spines[["top", "right"]].set_visible(False)
plt.tight_layout(); plt.show()
print(score.round(2).to_string())
```

```python
COUNT_FEATS = ["modis_n", "modis_terra_n", "modis_night_n", "nb_modis_n", "modis_n_prev",
               "moy", "lat", "lon", "borneo", "cell_ratio", "cell_viirs_rate", "modis_x_ratio"]
FRP_FEATS   = ["frp_sum", "frp_mean", "frp_max", "nb_frp_sum", "frp_sum_prev", "frp_per_hotspot_nb"]
FEATURE_SETS = {"counts only": COUNT_FEATS, "counts + FRP grid": COUNT_FEATS + FRP_FEATS}

NO_LOG = {"lat", "lon", "borneo", "cell_ratio", "moy"}

def make_X(d, feats, kind):
    X = d[feats].astype(float)
    if kind in ("linear", "poisson"):
        X = X.drop(columns="moy").join(pd.get_dummies(d.moy, prefix="m", dtype=float))
    if kind == "poisson":
        heavy = [c for c in feats if c not in NO_LOG]
        X[heavy] = np.log1p(X[heavy])
    return X

MODELS = {
    "Linear regression":          ("linear",  lambda: LinearRegression()),
    "Poisson GLM":                ("poisson", lambda: make_pipeline(StandardScaler(), PoissonRegressor(alpha=1e-3, max_iter=3000))),
    "Random forest":              ("tree",    lambda: RandomForestRegressor(n_estimators=300, min_samples_leaf=5, max_features=0.5,
                                                                            n_jobs=-1, random_state=0)),
    "Gradient boosting (squared)":("tree",    lambda: HistGradientBoostingRegressor(loss="squared_error", learning_rate=0.05, max_iter=600,
                                                                                    min_samples_leaf=40, l2_regularization=1.0, random_state=0)),
    "Gradient boosting (Poisson)":("tree",    lambda: HistGradientBoostingRegressor(loss="poisson", learning_rate=0.05, max_iter=600,
                                                                                    min_samples_leaf=40, l2_regularization=1.0, random_state=0)),
}

def evaluate(d, pred):
    d = d.assign(pred=np.clip(pred, 0, None))
    nat = d.groupby("month")[["pred", "viirs"]].sum()
    reg = d.groupby(["month", "borneo"])[["pred", "viirs"]].sum()
    act = (d.viirs > 0) | (d.modis_n > 0)
    e = nat.pred - nat.viirs
    return {
        "MAE national/month":   mean_absolute_error(nat.viirs, nat.pred),
        "MAE region/month":     mean_absolute_error(reg.viirs, reg.pred),
        "MAE cell/month":       mean_absolute_error(d.viirs, d.pred),
        "MAE active cells":     mean_absolute_error(d.viirs[act], d.pred[act]),
        "accuracy_% national":  100 * (1 - e.abs().sum() / nat.viirs.sum()),
        "R2 national":          1 - (e ** 2).sum() / ((nat.viirs - nat.viirs.mean()) ** 2).sum(),
        "bias_%":               100 * e.sum() / nat.viirs.sum(),
    }

def run_all(p, k0, models=MODELS, feature_sets=FEATURE_SETS):
    tr, te = p[p.period == "learn"], p[p.period == "test"]
    results, preds, fitted = [], {}, {}
    for name, pr in [("Ratio - single k (current method)", k0 * te.modis_n),
                     ("Ratio - per cell",                  te.modis_x_ratio)]:
        results.append({"method": name, "features": "count", **evaluate(te, pr)})
        preds[(name, "count")] = np.asarray(pr)
    for mname, (kind, factory) in models.items():
        for fname, feats in feature_sets.items():
            model = factory().fit(make_X(tr, feats, kind), tr.viirs)
            pr = model.predict(make_X(te, feats, kind))
            results.append({"method": mname, "features": fname, **evaluate(te, pr)})
            preds[(mname, fname)] = np.clip(pr, 0, None)
            fitted[(mname, fname)] = model
            print(f"done: {mname:30s} | {fname:18s} | national MAE {results[-1]['MAE national/month']:.1f}")
    res = pd.DataFrame(results).sort_values("MAE national/month").reset_index(drop=True)
    return res, preds, fitted

res, preds, fitted = run_all(panel, k0)
print()
print(res.round(2).to_string(index=False))
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing
import warnings

last_day  = harm.acq_date.max()
lm        = last_day.to_period("M").to_timestamp()
LAST_FULL = lm if (last_day + pd.Timedelta(days=1)).month != last_day.month else lm - pd.offsets.MonthBegin(1)
y = series[series.index <= LAST_FULL].asfreq("MS", fill_value=0)
print(f"harmonized series: {y.index.min().date()} to {y.index.max().date()} ({len(y)} months; last data day {last_day.date()})")

def _future(y, h): return pd.date_range(y.index[-1] + pd.offsets.MonthBegin(1), periods=h, freq="MS")

def fc_clim(y, h, years=None):
    hist = y if years is None else y[y.index >= y.index[-1] - pd.DateOffset(years=years)]
    med = hist.groupby(hist.index.month).median()
    idx = _future(y, h); return pd.Series(med.reindex(idx.month).values, idx)

def fc_last_year(y, h):
    idx = _future(y, h); return pd.Series([y[d - pd.DateOffset(years=1 + i // 12)] for i, d in enumerate(idx)], idx)

def fc_hw(y, h):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        m = ExponentialSmoothing(np.log1p(y), trend=None, seasonal="add", seasonal_periods=12).fit()
    return np.expm1(m.forecast(h)).clip(lower=0)

def fc_sarima(y, h):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        m = SARIMAX(np.log1p(y), order=(1, 0, 1), seasonal_order=(1, 1, 1, 12)).fit(disp=False)
    return np.expm1(m.forecast(h)).clip(lower=0)

def fc_gbm(y, h):
    z, moy = list(np.log1p(y.values)), list(y.index.month)
    f = lambda t, mo: [z[t-1], z[t-2], z[t-3], z[t-12], z[t-24], np.mean(z[t-12:t]), mo]
    X = [f(t, moy[t]) for t in range(24, len(z))]
    m = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, min_samples_leaf=10, random_state=0).fit(X, z[24:])
    idx = _future(y, h)
    for mo in idx.month:
        z.append(float(m.predict([f(len(z), mo)])[0]))
    return pd.Series(np.expm1(z[-h:]), idx).clip(lower=0)

def fc_ensemble(y, h):
    parts = [fc_sarima(y, h), fc_hw(y, h), fc_clim(y, h)]
    return pd.Series(np.expm1(np.mean([np.log1p(p.values) for p in parts], axis=0)), parts[0].index)

FORECASTERS = {
    "Climatology (all years)":   fc_clim,
    "Climatology (last 10 yrs)": lambda y, h: fc_clim(y, h, years=10),
    "Same month last year":      fc_last_year,
    "Holt-Winters (log)":        fc_hw,
    "SARIMA (log)":              fc_sarima,
    "Gradient boosting on lags": fc_gbm,
    "Ensemble (SARIMA+HW+clim)": fc_ensemble,
}

H = 12
ORIGINS = [LAST_FULL - pd.DateOffset(months=12 * k) for k in (4, 3, 2, 1)]
bt = []
for name, f in FORECASTERS.items():
    for o in ORIGINS:
        fc = f(y[y.index <= o], H)
        act = y.reindex(fc.index)
        bt.append(pd.DataFrame({"method": name, "origin": o, "month": fc.index, "pred": fc.values, "actual": act.values}))
bt = pd.concat(bt, ignore_index=True)
bt["abs_err"] = (bt.pred - bt.actual).abs()
bt["log_err"] = np.log1p(bt.actual) - np.log1p(bt.pred)

fc_res = bt.groupby("method").apply(lambda d: pd.Series({
    "MAE (hotspots/month)": d.abs_err.mean(),
    "median abs err":       d.abs_err.median(),
    "bias_%":               100 * (d.pred.sum() - d.actual.sum()) / d.actual.sum(),
    **{f"MAE from {o:%Y-%m}": d[d.origin == o].abs_err.mean() for o in ORIGINS},
}), include_groups=False).sort_values("MAE (hotspots/month)")
print(fc_res.round(1).to_string())
best_fc = fc_res.index[0]
print("\nbest forecasting method:", best_fc)
```

```python
import requests, time
POWER_URL = "https://power.larc.nasa.gov/api/temporal/daily/point"
HAZE_DIR = f"{BASE}/haze"; os.makedirs(HAZE_DIR, exist_ok=True)
W_START, W_END = "20001101", (HZ_END + pd.offsets.MonthEnd(0)).strftime("%Y%m%d")

WX_POINTS = {
    "Kuala Lumpur":                  (3.14, 101.69, "Peninsular"),
    "Johor Bahru":                   (1.49, 103.74, "Peninsular"),
    "Kuching":                       (1.55, 110.35, "Borneo"),
    "Kota Kinabalu":                 (5.98, 116.07, "Borneo"),
    "Pekanbaru (Riau)":              (0.51, 101.45, "Sumatra"),
    "Palembang (S. Sumatra)":        (-2.99, 104.76, "Sumatra"),
    "Pontianak (W. Kalimantan)":     (-0.03, 109.33, "Kalimantan"),
    "Palangka Raya (C. Kalimantan)": (-2.21, 113.92, "Kalimantan"),
}

def power_daily(name, lat, lon, params, community):
    cache = f"{HAZE_DIR}/power_{name.split(' ')[0].lower()}_{params.replace(',', '-')}_{W_START}_{W_END}.csv"
    if os.path.exists(cache):
        return pd.read_csv(cache, parse_dates=["date"])
    r = requests.get(POWER_URL, timeout=300, params={"parameters": params, "community": community, "latitude": lat,
                     "longitude": lon, "start": W_START, "end": W_END, "format": "JSON"})
    r.raise_for_status()
    d = pd.DataFrame(r.json()["properties"]["parameter"])
    d.index = pd.to_datetime(d.index, format="%Y%m%d")
    d = d.replace(-999.0, np.nan).rename_axis("date").reset_index()
    d.to_csv(cache, index=False); time.sleep(1)
    return d

def to_monthly(d):
    d = d.set_index("date")
    rad = np.deg2rad(d.WD10M)
    d["u"], d["v"] = -d.WS10M * np.sin(rad), -d.WS10M * np.cos(rad)
    d["dry_day"] = (d.PRECTOTCORR < 1).astype(float)
    return d.resample("MS").agg({"T2M": "mean", "RH2M": "mean", "PRECTOTCORR": "sum", "WS10M": "mean",
                                 "u": "mean", "v": "mean", "dry_day": "sum"}).rename(columns={"PRECTOTCORR": "rain_mm"})

wx_pts = {}
for name, (lat, lon, grp) in WX_POINTS.items():
    wx_pts[name] = to_monthly(power_daily(name, lat, lon, "T2M,RH2M,PRECTOTCORR,WS10M,WD10M", "AG")).assign(group=grp)
    print("weather ok:", name)
wx_all = pd.concat(wx_pts.values())
wx_m = {grp: d.drop(columns="group").groupby(level=0).mean().reindex(HZ_MONTHS) for grp, d in wx_all.groupby("group")}
CENT = {grp: tuple(np.mean([(la, lo) for la, lo, g_ in WX_POINTS.values() if g_ == grp], axis=0)) for grp in wx_m}

try:
    aod = []
    for name, (lat, lon, grp) in WX_POINTS.items():
        if grp in ("Peninsular", "Borneo"):
            a = power_daily(name, lat, lon, "AOD_55", "RE").set_index("date").AOD_55.resample("MS").mean()
            aod.append(pd.DataFrame({"region": grp, "month": a.index, "aod": a.to_numpy()}))
    aod_m = pd.concat(aod).groupby(["region", "month"]).aod.mean()
    print("AOD ok:", aod_m.dropna().index.get_level_values("month").min().date(), "to",
          aod_m.dropna().index.get_level_values("month").max().date())
except Exception as e:
    aod_m = None; print("AOD not available, skipping that check:", e)

print(wx_m["Peninsular"].groupby(wx_m["Peninsular"].index.month).mean().round(1))
```

```python
RECEPTORS = ["Peninsular", "Borneo"]
HAZE_W = {r: 0.4 for r in RECEPTORS}
Q_MOD, Q_SEV = 0.70, 0.90
LEVELS = ["Low", "Moderate", "Severe"]

def bearing(a, b):
    la1, lo1, la2, lo2 = map(np.radians, (a[0], a[1], b[0], b[1]))
    x = np.sin(lo2 - lo1) * np.cos(la2)
    y = np.cos(la1) * np.sin(la2) - np.sin(la1) * np.cos(la2) * np.cos(lo2 - lo1)
    return (np.degrees(np.arctan2(x, y)) + 360) % 360

def dist_km(a, b):
    la1, lo1, la2, lo2 = map(np.radians, (a[0], a[1], b[0], b[1]))
    h = np.sin((la2 - la1) / 2) ** 2 + np.cos(la1) * np.cos(la2) * np.sin((lo2 - lo1) / 2) ** 2
    return 2 * 6371 * np.arcsin(np.sqrt(h))

def transport(w, src, rec):
    b = np.radians(bearing(CENT[src], CENT[rec]))
    along = w.u * np.sin(b) + w.v * np.cos(b)
    steady = (along / w.WS10M.replace(0, np.nan)).clip(0, 1).fillna(0)
    return (0.1 + 0.9 * steady) * min(1.0, 300.0 / dist_km(CENT[src], CENT[rec]))

for s_, r_ in [(s, r) for s in SRC for r in RECEPTORS]:
    print(f"{s_:10s} -> {r_:10s}: bearing {bearing(CENT[s_], CENT[r_]):5.0f} deg, {dist_km(CENT[s_], CENT[r_]):5.0f} km")

RAIN_NORM = {grp: wx_m[grp].rain_mm.groupby(wx_m[grp].index.month).median() for grp in wx_m}

def build_drivers(local_m, ind_m, wx_m, months):
    rows = []
    for rec in RECEPTORS:
        d = pd.DataFrame(index=months)
        d["region"], d["borneo"] = rec, int(rec == "Borneo")
        d["local"] = local_m[rec].reindex(months).to_numpy()
        for src, s in SRC.items():
            ws = wx_m[src].reindex(months)
            d[f"{s}_n"], d[f"{s}_frp"] = ind_m[f"{s}_n"].reindex(months).to_numpy(), ind_m[f"{s}_frp"].reindex(months).to_numpy()
            d[f"T_{s}"] = transport(ws, src, rec).to_numpy()
            d[f"src_rain_{s}"] = ws.rain_mm.to_numpy()
            d[f"src_rain_anom_{s}"] = ws.rain_mm.to_numpy() - RAIN_NORM[src].reindex(months.month).to_numpy()
        d["transported_frp"] = d.sum_frp * d.T_sum + d.kal_frp * d.T_kal
        w = wx_m[rec].reindex(months)
        for c in ["rain_mm", "RH2M", "T2M", "WS10M", "dry_day"]:
            d[c] = w[c].to_numpy()
        d["rain_anom"] = d.rain_mm - RAIN_NORM[rec].reindex(months.month).to_numpy()
        rows.append(d)
    D = pd.concat(rows).rename_axis("month").reset_index()
    D["moy"], D["year"] = D.month.dt.month, D.month.dt.year
    D["moy_sin"], D["moy_cos"] = np.sin(2 * np.pi * D.moy / 12), np.cos(2 * np.pi * D.moy / 12)
    for c in ["local", "sum_n", "kal_n", "sum_frp", "kal_frp", "transported_frp"]:
        D["log_" + c] = np.log1p(D[c].clip(lower=0))
    return D

hz = build_drivers(local_m, ind_m, wx_m, HZ_MONTHS)

HSTATS = {r: {c: (hz.loc[hz.region == r, c].mean(), hz.loc[hz.region == r, c].std())
              for c in ["log_local", "log_transported_frp"]} for r in RECEPTORS}
def haze_parts(D):
    zl, zt = np.full(len(D), np.nan), np.full(len(D), np.nan)
    for r in RECEPTORS:
        m = (D.region == r).to_numpy(); st = HSTATS[r]
        zl[m] = (D.log_local[m] - st["log_local"][0]) / st["log_local"][1]
        zt[m] = (D.log_transported_frp[m] - st["log_transported_frp"][0]) / st["log_transported_frp"][1]
    return zl, zt

def haze_index(D):
    zl, zt = haze_parts(D); w = D.region.map(HAZE_W).to_numpy()
    return w * zl + (1 - w) * zt

spear = lambda x, y: pd.Series(x).corr(pd.Series(y), method="spearman")
if aod_m is not None:
    a = hz.merge(aod_m.reset_index(), on=["region", "month"]).dropna(subset=["aod"])
    zl, zt = haze_parts(a); trows = []
    for r in RECEPTORS:
        m = (a.region == r).to_numpy(); cal = m & (a.year <= 2012).to_numpy(); val = m & (a.year >= 2013).to_numpy()
        for w_ in np.round(np.arange(0, 0.8, 0.1), 1):
            hi_ = w_ * zl + (1 - w_) * zt
            trows.append({"region": r, "local_weight": w_, "AOD corr 2001-2012 (tuning)": spear(hi_[cal], a.aod.to_numpy()[cal]),
                          "AOD corr 2013+ (check)": spear(hi_[val], a.aod.to_numpy()[val])})
    tune = pd.DataFrame(trows)
    best_w = tune.loc[tune.groupby("region")["AOD corr 2001-2012 (tuning)"].idxmax()]
    HAZE_W = dict(zip(best_w.region, best_w.local_weight))
    print(tune.round(3).to_string(index=False)); print("\nchosen local weight per region:", HAZE_W)

hz["haze_index"] = haze_index(hz)
THRESH = {r: hz.loc[hz.region == r, "haze_index"].quantile([Q_MOD, Q_SEV]).to_numpy() for r in RECEPTORS}
def level_of(D, hi):
    t = np.vstack(D.region.map(THRESH).to_numpy())
    return (hi >= t[:, 0]).astype(int) + (hi >= t[:, 1]).astype(int)
hz["level"] = level_of(hz, hz.haze_index.to_numpy())
hz["level_name"] = hz.level.map(dict(enumerate(LEVELS)))
print(hz.groupby(["region", "level_name"]).size().unstack())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import f1_score, roc_auc_score

FUT_YEAR = HZ_END.year + 1
LMAX = (FUT_YEAR + 1 - HZ_END.year) * 12 + 12 - HZ_END.month
LEVEL = {r: hz[hz.region == r].set_index("month").level for r in RECEPTORS}

def anom_log(s):
    s = s.clip(lower=0); return np.log1p(s) - np.log1p(s.groupby(s.index.month).transform("median"))

ind_feat = pd.DataFrame(index=HZ_MONTHS)
for c in ["sum_frp", "kal_frp", "sumE_frp"]:
    ind_feat[f"{c}_anom"]  = anom_log(ind_m[c])
    ind_feat[f"{c}_anom3"] = anom_log(ind_m[c].rolling(3, min_periods=1).sum())
for s in ["sum", "kal"]:
    n_ = ind_m[f"{s}_n"].replace(0, np.nan)
    ind_feat[f"{s}_intensity"] = np.log1p(ind_m[f"{s}_frp"] / n_)
    ind_feat[f"{s}_night"]     = ind_m[f"{s}_night"] / n_
ind_feat = ind_feat.fillna(ind_feat.median())

STATE = {}
for r in RECEPTORS:
    d = hz[hz.region == r].set_index("month").reindex(HZ_MONTHS)
    st = pd.DataFrame(index=HZ_MONTHS)
    st["local_anom"], st["local_anom3"] = anom_log(d.local), anom_log(d.local.rolling(3, min_periods=1).sum())
    st["trans_anom"] = anom_log(d.transported_frp)
    st["rain_anom3"] = d.rain_anom.rolling(3, min_periods=1).sum()
    st["src_rain_anom3_sum"] = d.src_rain_anom_sum.rolling(3, min_periods=1).sum()
    st["src_rain_anom3_kal"] = d.src_rain_anom_kal.rolling(3, min_periods=1).sum()
    st["hi_now"], st["level_now"] = d.haze_index, d.level
    st["issue_sin"], st["issue_cos"] = np.sin(2 * np.pi * st.index.month / 12), np.cos(2 * np.pi * st.index.month / 12)
    STATE[r] = st.join(ind_feat).join(clim[["nino", "nino_chg3", "dmi", "dmi_chg3"]])

def make_rows(r, issues, leads):
    st, out = STATE[r].reindex(issues), []
    for L in leads:
        X = st.copy(); tgt = pd.DatetimeIndex(issues) + pd.DateOffset(months=L)
        X["issue"], X["target"], X["lead"] = issues, tgt, L
        X["tgt_sin"], X["tgt_cos"] = np.sin(2 * np.pi * tgt.month / 12), np.cos(2 * np.pi * tgt.month / 12)
        X["nino_t"], X["dmi_t"] = clim.nino.reindex(tgt).to_numpy(), clim.dmi.reindex(tgt).to_numpy()
        X["level_t"] = LEVEL[r].reindex(tgt).to_numpy()
        out.append(X)
    return pd.concat(out, ignore_index=True).assign(region=r, borneo=int(r == "Borneo"))

ML = pd.concat([make_rows(r, HZ_MONTHS[3:], range(1, LMAX + 1)) for r in RECEPTORS], ignore_index=True)
ML = ML.dropna(subset=["level_t"]).reset_index(drop=True); ML["level_t"] = ML.level_t.astype(int)
print(f"forecast training rows: {len(ML):,} (2 regions x issue months x leads 1-{LMAX})")

F_TGT    = ["tgt_sin", "tgt_cos", "lead", "borneo"]
F_NOW    = ["issue_sin", "issue_cos", "hi_now", "level_now", "local_anom", "local_anom3", "trans_anom", "rain_anom3",
            "src_rain_anom3_sum", "src_rain_anom3_kal", "sum_frp_anom", "sum_frp_anom3", "kal_frp_anom", "kal_frp_anom3"]
F_INT    = ["sumE_frp_anom", "sumE_frp_anom3", "sum_intensity", "kal_intensity", "sum_night", "kal_night"]
F_ENSO   = ["nino", "nino_chg3", "dmi", "dmi_chg3"]
F_ENSO_T = ["nino_t", "dmi_t"]
LEAD_SETS = {"season only":                    F_TGT,
             "+ fires & weather now":          F_TGT + F_NOW,
             "+ Indonesian fire intensity":    F_TGT + F_NOW + F_INT,
             "+ ENSO & IOD now":               F_TGT + F_NOW + F_INT + F_ENSO}
SCEN_FEATS = LEAD_SETS["+ ENSO & IOD now"] + F_ENSO_T

LEAD_MODELS = {
    "Gradient boosting":   lambda: HistGradientBoostingClassifier(learning_rate=0.05, max_iter=250, max_depth=3,
                                                                  min_samples_leaf=40, l2_regularization=1.0, random_state=0),
    "Logistic regression": lambda: make_pipeline(SimpleImputer(strategy="median"), StandardScaler(),
                                                 LogisticRegression(max_iter=5000, C=0.5)),
    "Random forest":       lambda: make_pipeline(SimpleImputer(strategy="median"),
                                                 RandomForestClassifier(n_estimators=300, min_samples_leaf=10, max_features=0.5,
                                                                        n_jobs=-1, random_state=0)),
}
```

```python
BT_YEARS = [y for y in range(2013, HZ_END.year) if (hz.year == y).sum() == 24]

def run_backtest(model, feats):
    out = []
    for Y in BT_YEARS:
        tr = ML[ML.target.dt.year < Y]; te = ML[ML.target.dt.year == Y]
        clf = LEAD_MODELS[model]().fit(tr[feats], tr.level_t)
        P = np.zeros((len(te), 3)); P[:, clf.classes_] = clf.predict_proba(te[feats])
        base = (tr.assign(moy=tr.target.dt.month).groupby(["region", "moy"]).level_t.value_counts(normalize=True)
                .unstack(fill_value=0).reindex(columns=[0, 1, 2], fill_value=0))
        B = base.reindex(pd.MultiIndex.from_arrays([te.region, te.target.dt.month])).fillna(1 / 3).to_numpy()
        out.append(te[["region", "issue", "target", "lead", "level_t"]].assign(
            P_low=P[:, 0], P_mod=P[:, 1], P_sev=P[:, 2], B_low=B[:, 0], B_mod=B[:, 1], B_sev=B[:, 2]))
    return pd.concat(out, ignore_index=True)

def rps(P, y):
    O = np.eye(3)[y]; return (((np.cumsum(P, 1) - np.cumsum(O, 1)) ** 2)[:, :2].sum(1) / 2).mean()

def warn_level(P, t_sev, t_mod):
    return np.where(P[:, 2] >= t_sev, 2, np.where(P[:, 1] + P[:, 2] >= t_mod, 1, 0))

def scores(d, pfx="P", t_sev=0.3, t_mod=0.5):
    P, y = d[[f"{pfx}_low", f"{pfx}_mod", f"{pfx}_sev"]].to_numpy(), d.level_t.to_numpy()
    wl = warn_level(P, t_sev, t_mod); warn = P[:, 2] >= t_sev
    return {"MAE (levels)":           np.abs(P[:, 1] + 2 * P[:, 2] - y).mean(),
            "MAE warning level":      np.abs(wl - y).mean(),
            "macro F1 (warning)":     f1_score(y, wl, average="macro", labels=[0, 1, 2], zero_division=0),
            "Severe caught_%":        100 * warn[y == 2].mean() if (y == 2).any() else np.nan,
            "Severe false alarms_%":  100 * (y[warn] != 2).mean() if warn.any() else np.nan,
            "Severe AUC":             roc_auc_score(y == 2, P[:, 2]) if 0 < (y == 2).sum() < len(y) else np.nan,
            "RPS":                    rps(P, y)}

bt_runs, rows_ = {}, []
for fname, feats in LEAD_SETS.items():
    bt_runs[("Gradient boosting", fname)] = run_backtest("Gradient boosting", feats); print("done:", fname)
for mname in ["Logistic regression", "Random forest"]:
    bt_runs[(mname, "+ ENSO & IOD now")] = run_backtest(mname, LEAD_SETS["+ ENSO & IOD now"]); print("done:", mname)
bt_scen = run_backtest("Gradient boosting", SCEN_FEATS); print("done: scenario model")

ref = next(iter(bt_runs.values()))
rows_.append({"model": "Baseline: usual level for the month", "features": "-", **scores(ref, "B")})
for (mname, fname), d in bt_runs.items():
    rows_.append({"model": mname, "features": fname, **scores(d)})
rows_.append({"model": "Gradient boosting", "features": "+ ENSO & IOD at target month (perfect ENSO outlook)", **scores(bt_scen)})
lead_res = pd.DataFrame(rows_)
lead_res["RPS skill vs baseline_%"] = 100 * (1 - lead_res.RPS / lead_res.RPS.iloc[0])
print("\nAll leads, both regions, 2013-2025 (Severe warning = P(Severe) >= 30%, same rule as the old backtest):")
print(lead_res.round(3).to_string(index=False))

best_key = min(bt_runs, key=lambda k: rps(bt_runs[k][["P_low", "P_mod", "P_sev"]].to_numpy(), bt_runs[k].level_t.to_numpy()))
best_bt = bt_runs[best_key]
print(f"\nbest forecast model: {best_key[0]} [{best_key[1]}]")
```

```python
GRID_T = np.round(np.arange(0.15, 0.75, 0.05), 2)
TAU = {}
for r in RECEPTORS:
    d = best_bt[best_bt.region == r]; P, y = d[["P_low", "P_mod", "P_sev"]].to_numpy(), d.level_t.to_numpy()
    t_sev = max(GRID_T, key=lambda t: f1_score(y == 2, P[:, 2] >= t, zero_division=0))
    t_mod = max(GRID_T, key=lambda t: f1_score(y >= 1, P[:, 1] + P[:, 2] >= t, zero_division=0))
    TAU[r] = (float(t_sev), float(t_mod))
print("warning thresholds (P(Severe), P(>=Moderate)):", TAU)

def table(d_all, label):
    rows = []
    for r in RECEPTORS:
        from_dec = (d_all.issue.dt.month == 12) & (d_all.target.dt.year == d_all.issue.dt.year + 1)
        for name, sel in [("all leads", d_all.lead > 0), ("1-3 months ahead", d_all.lead <= 3),
                          ("4-6 months ahead", d_all.lead.between(4, 6)), ("7-12 months ahead", d_all.lead.between(7, 12)),
                          ("Jan-Dec from previous December (old setup)", from_dec)]:
            d = d_all[(d_all.region == r) & sel]
            s, b = scores(d, "P", *TAU[r]), scores(d, "B", *TAU[r])
            rows.append({"forecast": label, "region": r, "subset": name, **s,
                         "RPS skill vs baseline_%": 100 * (1 - s["RPS"] / b["RPS"])})
    return pd.DataFrame(rows)

bt_table = pd.concat([table(best_bt, f"{best_key[0]} [{best_key[1]}]"),
                      table(bt_scen, "scenario model (ENSO outlook correct)")], ignore_index=True)
print(bt_table.round(3).to_string(index=False))

fig, axes = plt.subplots(2, 1, figsize=(13, 6), sharex=True)
for ax, r in zip(axes, RECEPTORS):
    d = best_bt[(best_bt.region == r) & (best_bt.lead == 3)].sort_values("target")
    ax.step(d.target, d.level_t, where="mid", color="black", lw=1, label="actual level")
    ax.plot(d.target, d.P_mod + 2 * d.P_sev, color="tab:blue", lw=1.6, label="forecast 3 months ahead (expected level)")
    ax.fill_between(d.target, 0, 2 * d.P_sev, color="tab:red", alpha=0.25, label="P(Severe) x 2")
    ax.set_yticks([0, 1, 2]); ax.set_yticklabels(LEVELS); ax.set_title(f"{r}: backtest, forecast made 3 months before")
    ax.spines[["top", "right"]].set_visible(False)
axes[0].legend(ncol=3, fontsize=8); plt.tight_layout(); plt.show()
```

```python
cell_lookup = cells.assign(cell_id=cells.cell_lat.map("{:.2f}".format) + "_" + cells.cell_lon.map("{:.2f}".format),
                           center_lat=cells.cell_lat + G_CAL / 2, center_lon=cells.cell_lon + G_CAL / 2)

def km_between(lat1, lon1, lat2, lon2):
    la1, lo1, la2, lo2 = map(np.radians, (lat1, lon1, lat2, lon2))
    h = np.sin((la2 - la1) / 2) ** 2 + np.cos(la1) * np.cos(la2) * np.sin((lo2 - lo1) / 2) ** 2
    return 2 * 6371 * np.arcsin(np.sqrt(h))

def select_cells(lat=None, lon=None, radius_km=None, box=None):
    c = cell_lookup
    if box is not None:
        sel = c.center_lat.between(box[0], box[1]) & c.center_lon.between(box[2], box[3])
    elif radius_km is not None:
        sel = km_between(lat, lon, c.center_lat, c.center_lon) <= radius_km
    else:
        cl, cn = to_cell(np.array([lat]), np.array([lon]))
        sel = (c.cell_lat == cl[0]) & (c.cell_lon == cn[0])
    return set(c.cell_id[sel])

def area_monthly(lat=None, lon=None, radius_km=None, box=None):
    ids = select_cells(lat, lon, radius_km, box)
    d = grid[grid.cell_id.isin(ids)]
    if d.empty:
        return pd.DataFrame()
    out = d.groupby("date").agg(squares=("cell_id", "nunique"), status=("status", "first"),
                                hotspots=("hotspots", "sum"), frp_total_mw=("frp_total_mw", "sum"),
                                haze_level=("haze_level", "max"), haze_p_severe=("haze_p_severe", "max"))
    out["frp_per_hotspot_mw"] = (out.frp_total_mw / out.hotspots.where(out.hotspots >= 0.5)).round(2)
    out["haze_level_name"] = out.haze_level.map(dict(enumerate(LEVELS)))
    mo = pd.to_datetime(out.index)
    out["moy"] = mo.month
    rec  = (out.status != "forecast").to_numpy()
    base = rec & (mo.year <= BASE_END)
    out["hotspots_median"]       = out.moy.map(out[base].groupby("moy").hotspots.median())
    out["hotspots_vs_median"]    = out.hotspots / out.hotspots_median.where(out.hotspots_median > 0)
    out["hotspots_minus_median"] = out.hotspots - out.hotspots_median
    out["rank_in_month"]   = out[rec].groupby("moy").hotspots.rank(ascending=False, method="min").astype("Int64")
    out["years_in_record"] = out[rec].groupby("moy").hotspots.transform("size").astype("Int64")
    out["hotspots_pct_rank"] = np.where(rec, below_share(out.reset_index(), ["moy"], "hotspots", base), np.nan)
    return out.drop(columns="moy").round(2)

print("50 km around Kuching:")
print(area_monthly(lat=1.55, lon=110.35, radius_km=50).tail(15).to_string())
```

```python
def area_hotspots(start, end, lat=None, lon=None, radius_km=None, box=None, exact=False, save_csv=None):
    start, end = str(pd.Timestamp(start).date()), str(pd.Timestamp(end).date())
    d = hot[(hot.date >= start) & (hot.date <= end)]
    if exact and radius_km is not None:
        d = d[km_between(lat, lon, d.latitude, d.longitude) <= radius_km]
    elif exact and box is not None:
        d = d[d.latitude.between(box[0], box[1]) & d.longitude.between(box[2], box[3])]
    else:
        d = d[d.cell_id.isin(select_cells(lat, lon, radius_km, box))]
    s = {"period": f"{start} to {end}", "detections (rows)": len(d),
         "hotspots (VIIRS-equivalent)": round(d.viirs_equivalent.sum(), 1),
         "of which converted MODIS": round(d.viirs_equivalent[d.sensor == "MODIS"].sum(), 1),
         "total FRP (MW, VIIRS-equiv)": round(d.frp_viirs_equiv_mw.sum(), 1),
         "days with fire": d.date.nunique(),
         "busiest day": (d.groupby("date").viirs_equivalent.sum().idxmax() if len(d) else None)}
    if start < str(cut_viirs.date()) and (pd.Timestamp(end) - pd.Timestamp(start)).days < 28:
        s["note"] = "short range before 2012: converted counts are calibrated per month, so treat this as rough"
    if save_csv:
        d.to_csv(save_csv, index=False)
    return d, s

def month_to_date(lat=None, lon=None, radius_km=None, box=None, as_of=None):
    as_of = pd.Timestamp(as_of or hot.date.max())
    ids = select_cells(lat, lon, radius_km, box)
    a = hot[hot.cell_id.isin(ids)]
    a = a.assign(ts=pd.to_datetime(a.date))
    def days_total(y):
        last = min(as_of.day, pd.Timestamp(y, as_of.month, 1).days_in_month)
        m = (a.ts.dt.year == y) & (a.ts.dt.month == as_of.month) & (a.ts.dt.day <= last)
        return a.viirs_equivalent[m].sum()
    first_year = 2000 if as_of.month >= 11 else 2001
    past = pd.Series({y: days_total(y) for y in range(first_year, as_of.year)}, name="hotspots")
    now = days_total(as_of.year)
    med = past.median()
    return {"period": f"{as_of:%Y-%m}-01 to {as_of:%Y-%m-%d}", "hotspots this year": round(now, 1),
            "median, same days, past years": round(med, 1),
            "vs median": (round(now / med, 2) if med > 0 else None),
            "rank": f"{int((past > now).sum()) + 1} of {len(past) + 1}",
            "share of past years with fewer": round((past < now).mean(), 2)}, past

rows_, summary_ = area_hotspots("2019-09-01", "2019-09-30", lat=2.9, lon=101.6, radius_km=50,
                                save_csv=f"{OUT}/example_search_2019-09_klang_valley.csv")
print("Klang Valley (50 km), September 2019:"); print(pd.Series(summary_).to_string())
mtd, past_ = month_to_date(lat=2.9, lon=101.6, radius_km=50)
print("\nKlang Valley (50 km), current month so far vs the same days in past years:"); print(pd.Series(mtd).to_string())
```
